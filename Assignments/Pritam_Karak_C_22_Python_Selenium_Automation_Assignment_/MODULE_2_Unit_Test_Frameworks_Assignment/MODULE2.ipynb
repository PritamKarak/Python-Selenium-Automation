# Assignment 7: Page Object Model (POM) Restructure

## Objective

The objective of this assignment is to restructure a Selenium automation script using the Page Object Model (POM) design pattern.

## Introduction

Page Object Model is a design pattern commonly used in Selenium automation. It separates the web page elements and their actions from the actual test cases.

Instead of writing all locators and Selenium commands directly inside the test script, we create separate Python classes for different pages.

For example, a LoginPage class can contain the username field, password field, login button and login method. The test case can then simply call the login() method without knowing the implementation details of the locators.

## Advantages of POM

- Improves code readability.
- Reduces code duplication.
- Makes locators easier to maintain.
- Allows the same page methods to be reused in multiple tests.
- Separates test logic from webpage implementation.

## Application Used

For this assignment, the SauceDemo website is used because it provides a simple login page and a dashboard after successful authentication.

Website: https://www.saucedemo.com/

## Project Structure

The project is organized as follows:

## Project Structure

### Module2

**1. pages**
- `login_page.py` — Contains login page locators and methods.
- `dashboard_page.py` — Contains dashboard page locators and methods.

**2. tests**
- `test_login.py` — Contains the login test case and assertions.
The LoginPage class contains the login page locators and methods.

The DashboardPage class contains the dashboard-related locators and methods.

The test file contains only the actual test logic and assertions.

## Login Page

The LoginPage class represents the login page of the application.

It contains locators for the username field, password field and login button.

The login() method combines the individual actions required to log into the application.

```python
from selenium.webdriver.common.by import By

class LoginPage:

    username = (By.ID, "user-name")
    password = (By.ID, "password")
    login_button = (By.ID, "login-button")

    def __init__(self, driver):
        self.driver = driver

    def login(self, username, password):
        self.driver.find_element(*self.username).send_keys(username)
        self.driver.find_element(*self.password).send_keys(password)
        self.driver.find_element(*self.login_button).click()
```

## Dashboard Page

The DashboardPage class represents the page displayed after successful login.

The Products heading is used to verify that the login was successful.

The get_title() method retrieves the text displayed on the dashboard.

```python
from selenium.webdriver.common.by import By

class DashboardPage:

    products_title = (By.CLASS_NAME, "title")

    def __init__(self, driver):
        self.driver = driver

    def get_title(self):
        return self.driver.find_element(*self.products_title).text
```

## Test Case

The test case creates the WebDriver and initializes the LoginPage and DashboardPage objects.

The test then calls the login() method with valid credentials.

After login, the dashboard title is checked using an assertion.

This demonstrates how the test logic remains separate from the page locators.

```python
from selenium import webdriver
import time

driver = webdriver.Chrome()

driver.get("https://www.saucedemo.com/")
driver.maximize_window()

time.sleep(2)

login_page = LoginPage(driver)
dashboard_page = DashboardPage(driver)

time.sleep(2)

login_page.login("standard_user", "secret_sauce")

time.sleep(3)

assert dashboard_page.get_title() == "Products"

print("Login test passed successfully.")

time.sleep(5)

driver.quit()
```

# Assignment 8: Data-Driven Automation (DDT)

## Objective

The objective of this assignment is to create a data-driven Selenium automation script that reads multiple test cases from an external data source.

## Introduction

Data-Driven Testing is an automation technique in which test data is stored separately from the test script.

Instead of creating a separate test script for every username and password combination, the same test script can be executed repeatedly using different data.

For this assignment, a CSV file is used as the external data source and the pandas library is used to read the CSV file.

The test contains valid and invalid username-password combinations.

## Advantages of Data-Driven Testing

- Allows multiple test cases to use the same test script.
- Reduces code duplication.
- Makes it easy to add new test data.
- Improves test coverage.
- Separates test data from automation logic.

## Application Used

SauceDemo is used as the login application.

Website: https://www.saucedemo.com/

The test checks both successful login and invalid login scenarios.

## External Test Data

The test data is stored in a CSV file named `login_data.csv`.

The CSV file contains three columns:

- username
- password
- expected

The expected column specifies whether the login should be successful or should display a validation error.

## Reading the External Data

The pandas library is used to read the CSV file.

The `read_csv()` function loads the test data into a DataFrame.

The test can then iterate through each row and use the username, password and expected result.

```python
import pandas as pd

data = pd.read_csv("login_data.csv")

print(data)
```

## Executing the Test Cases

The Selenium script loops through every row in the CSV file.

For each test case:

1. Chrome is opened.
2. The SauceDemo login page is opened.
3. The username is entered.
4. The password is entered.
5. The Login button is clicked.
6. The expected result is checked.
7. The browser is closed.
8. The next test case is executed.

This allows multiple combinations of valid and invalid credentials to be tested using one automation script.

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time

data = pd.read_csv("login_data.csv")

for index, row in data.iterrows():

    driver = webdriver.Chrome()

    driver.get("https://www.saucedemo.com/")
    driver.maximize_window()

    time.sleep(2)

    driver.find_element(
        By.ID, "user-name"
    ).send_keys(row["username"])

    time.sleep(1)

    driver.find_element(
        By.ID, "password"
    ).send_keys(row["password"])

    time.sleep(1)

    driver.find_element(
        By.ID, "login-button"
    ).click()

    time.sleep(3)

    if row["expected"] == "success":

        title = driver.find_element(
            By.CLASS_NAME, "title"
        ).text

        assert title == "Products"

        print(
            f"Test {index + 1}: Login successful"
        )

    else:

        error = driver.find_element(
            By.CSS_SELECTOR,
            "h3[data-test='error']"
        ).text

        assert error != ""

        print(
            f"Test {index + 1}: Validation error displayed"
        )

    time.sleep(3)

    driver.quit()

    time.sleep(2)
```

## Expected Result

The script executes all the test cases stored in the CSV file.

For the valid credentials, the Products page is displayed.

For invalid credentials, an appropriate validation error is displayed.

Example output:

Test 1: Login successful
Test 2: Validation error displayed
Test 3: Validation error displayed
Test 4: Validation error displayed

Therefore, the same Selenium test script is successfully used with multiple external test data sets.

## Conclusion

Data-Driven Testing allows a single Selenium test script to execute multiple test scenarios using external data.

Using pandas and CSV makes the test data easy to modify without changing the Selenium automation code.

This approach improves reusability, maintainability and test coverage.

# Assignment 9: PyTest Integration with HTML Reporting

## Objective

The objective of this assignment is to integrate Selenium automation with PyTest and generate an HTML execution report.

## Introduction

PyTest is a Python testing framework that makes it easier to organize, execute and report automated tests.

PyTest fixtures can be used to initialize and close the Selenium WebDriver.

A fixture can open the browser before a test starts and close it after the test finishes.

The pytest-html plugin can be used to generate an HTML report containing the test execution results.

The framework can also capture screenshots when a test fails.

## Advantages

- Provides a structured testing framework.
- Supports fixtures for browser setup and teardown.
- Provides assertions.
- Supports multiple test cases.
- Generates HTML execution reports.
- Can capture screenshots for failed tests.

## PyTest Fixture

A PyTest fixture is used to manage the Selenium WebDriver.

The fixture starts Chrome before the test and provides the driver object to the test.

The `yield` statement separates setup and teardown.

The code before `yield` runs before the test.

The code after `yield` runs after the test.

Therefore, `driver.quit()` ensures that the browser is closed after the test finishes.

```python
import pytest
from selenium import webdriver

@pytest.fixture
def driver():

    driver = webdriver.Chrome()
    driver.maximize_window()

    yield driver

    driver.quit()
```

## Test Case

## HTML Reporting

## Capturing Screenshots for Failed Tests

To improve the HTML report, a screenshot can be captured whenever a test fails.

PyTest provides the `pytest_runtest_makereport` hook, which allows the framework to determine whether a test has passed or failed.

If the test fails, Selenium saves a screenshot of the current browser state.

The screenshot is then added to the pytest-html report.

```python
import pytest
from selenium import webdriver

@pytest.hookimpl(hookwrapper=True)
def pytest_runtest_makereport(item, call):

    outcome = yield
    report = outcome.get_result()

    setattr(item, "rep_" + report.when, report)


@pytest.fixture
def driver(request):

    driver = webdriver.Chrome()
    driver.maximize_window()

    yield driver

    if request.node.rep_call.failed:

        screenshot = f"{request.node.name}.png"

        driver.save_screenshot(screenshot)

        pytest_html = request.config.pluginmanager.getplugin("html")

        if pytest_html:

            extra = getattr(request.node, "extras", [])

            extra.append(
                pytest_html.extras.image(screenshot)
            )

            request.node.extras = extra

    driver.quit()
```
