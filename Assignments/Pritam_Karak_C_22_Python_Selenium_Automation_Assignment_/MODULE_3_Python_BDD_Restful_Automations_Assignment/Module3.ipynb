## Introduction to BDD

Behavior Driven Development (BDD) is a software development approach where the behavior of an application is described using simple, human-readable language.

Behave is a Python BDD framework that uses the Gherkin language.

The main Gherkin keywords are:

- Feature
- Scenario
- Given
- When
- Then
- And
- Scenario Outline
- Examples

## Gherkin Example

The following scenario describes a login test:

Feature: Login

Scenario: Successful login

Given the user opens the login page  
When the user enters valid username  
And the user enters valid password  
And the user clicks the login button  
Then the products page should be displayed

## Behave Project Structure

A typical Behave project is organized as follows:

module3_lab/

    features/
        login.feature
        steps/
            login_steps.py

    pages/
        login_page.py

## Day 1: QEMU and GDB Debugging

QEMU is used to simulate an embedded platform so that the application can be tested without requiring the actual hardware.

GDB is used for debugging the program running on the simulated platform.

### General Workflow

Host Computer
       ↓
     QEMU
       ↓
Simulated Embedded Platform
       ↓
    Program
       ↓
     GDB
       ↓
Breakpoints / Debugging

### GDB Debugging Steps

1. Start the QEMU simulated platform.
2. Start the application on the simulated platform.
3. Connect GDB to the simulated target.
4. Set a breakpoint.
5. Start program execution.
6. Inspect variables and program execution.
7. Continue execution and observe the result.

### Observation

QEMU provides a simulated environment for the embedded platform, while GDB provides debugging capabilities such as breakpoints and variable inspection.

# Day 3 – Debugging the Scenario Using Eclipse and GDB

## Objective

To debug the application running on the simulated embedded platform using Eclipse and GDB.

### Procedure

1. Open the project in Eclipse.
2. Configure the GDB debugger.
3. Connect Eclipse/GDB to the QEMU target.
4. Set a breakpoint at the required section of the program.
5. Start debugging.
6. Execute the program step by step.
7. Inspect variables and program execution.
8. Continue execution until the scenario is completed.

### Debugging Operations

The following GDB operations can be used:

- Breakpoint – Stop execution at a specific line.
- Step Over – Execute the current line and move to the next line.
- Step Into – Enter a function being called.
- Continue – Continue program execution.
- Variable Inspection – Check the current value of variables.

### Observation

The program execution was controlled using Eclipse and GDB. Breakpoints were used to inspect the execution flow and variables.

## Data-Driven Testing

Data-driven testing is a testing approach where the same test scenario is executed with different sets of input data.

In Behave, data-driven testing can be implemented using the `Scenario Outline` and `Examples` keywords.

### Advantages

- Reduces duplicate test cases.
- Allows multiple inputs to be tested.
- Makes test cases easier to maintain.
- Separates test data from test logic.

## Result
