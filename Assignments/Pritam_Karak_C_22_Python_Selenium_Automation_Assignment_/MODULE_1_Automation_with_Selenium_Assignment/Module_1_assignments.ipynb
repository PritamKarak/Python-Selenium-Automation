## Assignment 1: The Multi-Locator Challenge
**Task:** Navigate to a login page (e.g., SauceDemo). Find and interact with the username field using `By.ID`, the password field using `By.NAME`, and the login button using `By.XPATH`.  
**Validation:** Assert that the resulting page URL contains `/inventory.html` after logging in.

* **What we will do:** We will automate a user authentication flow and verify the post-login destination URL.
* **How we will do it:**
  * First, open SauceDemo's portal and locate target elements using ID, Name, and XPath selectors.
  * Enter the login credentials, trigger the login action, and assert that `/inventory.html` is in the current URL.


```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service as ChromeService
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.support.select import Select
import time

driver = webdriver.Chrome(service=ChromeService(ChromeDriverManager().install()))
driver.maximize_window()
driver.get("https://www.saucedemo.com/")

# Locate elements using specified locator strategies
driver.find_element(By.ID, "user-name").send_keys("standard_user")
driver.find_element(By.NAME, "password").send_keys("secret_sauce")
driver.find_element(By.XPATH, "//input[@id='login-button']").click()


time.sleep(5)
# Validate successful redirection
assert "/inventory.html" in driver.current_url, f"Login failed. URL: {driver.current_url}"
print("Validation Successful.")

driver.quit()
```

## Assignment 2: Synchronization & Explicit Waits
**Task:** Navigate to a page with dynamic content (e.g., an AJAX loader or a "Start" button that reveals text after 5 seconds).  
**Constraint:** Do not use `time.sleep()`. Implement `WebDriverWait` along with `expected_conditions` to wait until dynamic element appears.

* **What we will do:** Handle asynchronously loaded content using non-blocking synchronization strategies.
* **How we will do it:**
  * Navigate to the dynamic loading demo page and trigger the loading sequence.
  * Apply `WebDriverWait` with `EC.visibility_of_element_located` to wait until the dynamic element appears.
  * Read and assert the revealed text content.

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service as ChromeService
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.support.select import Select
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

driver = webdriver.Chrome(service=ChromeService(ChromeDriverManager().install()))
driver.maximize_window()
driver.get("https://the-internet.herokuapp.com/dynamic_loading/2")

driver.find_element(By.CSS_SELECTOR, "#start button").click()

# Explicit wait for dynamic text visibility
wait = WebDriverWait(driver, 10)
finish_element = wait.until(EC.visibility_of_element_located((By.CSS_SELECTOR, "#finish h4")))
print(driver.find_element(By.CSS_SELECTOR,"#finish h4").text)

assert finish_element.text == "Hello World!"
print("Dynamic text successfully validated.")

driver.quit()
```

## Assignment 4: JavaScript Alerts and Confirms
**Task:** Trigger a JavaScript Alert, a Confirm Box, and a Prompt Box on a practice page.  
**Task Steps:** Accept the first alert, dismiss the confirm box, and use `driver.switch_to.alert.send_keys()` to input text into the prompt before submitting.

* **What we will do:** Interact with and handle browser JavaScript modal dialogues.
* **How we will do it:**
  * Trigger each modal on the test page sequentially.
  * Switch execution focus to alerts using `driver.switch_to.alert`.
  * Accept the simple alert, dismiss the confirmation box, and enter text before accepting the prompt modal.

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service as ChromeService
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.support.select import Select
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

driver = webdriver.Chrome(service=ChromeService(ChromeDriverManager().install()))
driver.maximize_window()
driver.get("https://the-internet.herokuapp.com/javascript_alerts")

wait = WebDriverWait(driver, 10)

# 1. Accept Alert
driver.find_element(By.XPATH, "//button[text()='Click for JS Alert']").click()
wait.until(EC.alert_is_present()).accept()

# 2. Dismiss Confirm
driver.find_element(By.XPATH, "//button[text()='Click for JS Confirm']").click()
wait.until(EC.alert_is_present()).dismiss()

# 3. Enter text and Accept Prompt
driver.find_element(By.XPATH, "//button[text()='Click for JS Prompt']").click()
prompt = wait.until(EC.alert_is_present())
prompt.send_keys("Selenium Automation")
prompt.accept()

assert "Selenium Automation" in driver.find_element(By.ID, "result").text
print("Assignment 4 Passed: All alerts handled.")

driver.quit()
```

## Assignment 5: The HTML Web Table Extractor
**Task:** Find a page containing a multi-column dynamic data table (e.g., a stock list or user directory).  
**Task Steps:** Write a script to iterate through rows and columns. Locate a specific row by matching a name string, and retrieve the corresponding value from the "Status" or "Price" column next to it.

* **What we will do:** Parse tabular elements and extract specific column values dynamically.
* **How we will do it:**
  * Locate table rows in the DOM.
  * Iterate through rows to find a matching target string in the name column.
  * Extract and assert the value from the target column belonging to the matched row.

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service as ChromeService
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.support.select import Select

driver = webdriver.Chrome(service=ChromeService(ChromeDriverManager().install()))
driver.maximize_window()
driver.get("https://the-internet.herokuapp.com/tables")

target_last_name = "Smith"
target_due_amount = None

# Scan table rows for matching record
for row in driver.find_elements(By.XPATH, "//table[@id='table1']/tbody/tr"):
    cols = row.find_elements(By.TAG_NAME, "td")
    if cols[0].text == target_last_name:
        target_due_amount = cols[3].text
        break

assert target_due_amount is not None, "Target row was not found."
print(f"Assignment 5 Passed: Found {target_last_name} with Due {target_due_amount}")

driver.quit()
```

## Assignment 6: Windows, Tabs, and Iframes
**Task:** Open a page that contains an embedded iframe and a button that opens a new browser tab.  
**Task Steps:** Use `driver.switch_to.frame()` to interact with elements inside the iframe. Then, click the tab button, use `driver.window_handles` to switch context to the new tab, grab its title, close it, and switch back to the main layout.

* **What we will do:** Manage frame switching and handle multiple browser window contexts.
* **How we will do it:**
  * Switch into an iframe via `driver.switch_to.frame()`, interact with inner content, and return to default content.
  * Open a new window, identify handles via `driver.window_handles`, switch contexts, close the new tab, and restore the main window handle.

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service as ChromeService
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.support.select import Select

driver = webdriver.Chrome(service=ChromeService(ChromeDriverManager().install()))
driver.maximize_window()

# Part 1: Iframe interaction
driver.get("https://the-internet.herokuapp.com/iframe")
driver.switch_to.frame("mce_0_ifr")
print("Iframe Text:", driver.find_element(By.ID, "tinymce").text)
driver.switch_to.default_content()

# Part 2: Window / Tab switching
driver.get("https://the-internet.herokuapp.com/windows")
original_window = driver.current_window_handle
driver.find_element(By.LINK_TEXT, "Click Here").click()

for handle in driver.window_handles:
    if handle != original_window:
        driver.switch_to.window(handle)
        break

print("New Window Title:", driver.title)
driver.close()
driver.switch_to.window(original_window)

print("Assignment 6 Passed: Context switched and restored successfully.")
driver.quit()
```
