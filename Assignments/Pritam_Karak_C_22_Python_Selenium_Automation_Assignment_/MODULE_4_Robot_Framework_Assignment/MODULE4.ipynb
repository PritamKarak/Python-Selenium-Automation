# Assignment 1: Basic Syntax and Keywords

Robot Framework uses keywords to perform automation actions.

Some commonly used keywords are:

- `Open Browser`
- `Maximize Browser Window`
- `Input Text`
- `Click Button`
- `Page Should Contain`
- `Page Should Contain Element`
- `Close Browser`

The following examples demonstrate basic Robot Framework syntax using SeleniumLibrary.

## Question 1: Open a Web Browser and Navigate to a URL

### Explanation

The `Open Browser` keyword opens the specified browser and navigates to the given URL.

The `Maximize Browser Window` keyword maximizes the browser.

The `Page Should Contain` keyword verifies that specific text is present on the webpage.

Finally, `Close Browser` closes the browser.

In this example, the SauceDemo website is opened and the presence of the text `Swag Labs` is verified.

```python
from pathlib import Path
import subprocess

module4 = Path("Module4_Robot_Framework")
module4.mkdir(exist_ok=True)

basic_syntax = """*** Settings ***
Library    SeleniumLibrary

*** Variables ***
${URL}       https://www.saucedemo.com/
${BROWSER}   Chrome

*** Test Cases ***
Open Browser And Navigate
    Open Browser    ${URL}    ${BROWSER}
    Maximize Browser Window
    Sleep    3s
    Page Should Contain    Swag Labs
    Sleep    3s
    Close Browser
"""

(module4 / "basic_syntax.robot").write_text(
    basic_syntax,
    encoding="utf-8"
)

result = subprocess.run(
    ["robot", "basic_syntax.robot"],
    cwd=module4,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print(result.stderr)
```

## Question 2: Use the Input Text Keyword

### Explanation

The `Input Text` keyword is used to enter text into a textbox or form field.

In this example, the SauceDemo username and password fields are filled using the `Input Text` keyword.

The `id` locator is used to identify the input fields.

```python
input_text_test = """*** Settings ***
Library    SeleniumLibrary

*** Variables ***
${URL}       https://www.saucedemo.com/
${BROWSER}   Chrome

*** Test Cases ***
Input Text In Form
    Open Browser    ${URL}    ${BROWSER}
    Maximize Browser Window
    Sleep    2s

    Input Text    id=user-name    standard_user
    Sleep    2s

    Input Text    id=password    secret_sauce
    Sleep    3s

    Close Browser
"""

(module4 / "input_text.robot").write_text(
    input_text_test,
    encoding="utf-8"
)

result = subprocess.run(
    ["robot", "input_text.robot"],
    cwd=module4,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print(result.stderr)
```

## Question 3: Verify the Presence of an Element

### Explanation

The `Page Should Contain Element` keyword checks whether a specific element exists on the current webpage.

In this example, the username input field is checked using its ID.

If the element exists, the test passes. Otherwise, Robot Framework marks the test as failed.

```python
element_test = """*** Settings ***
Library    SeleniumLibrary

*** Variables ***
${URL}       https://www.saucedemo.com/
${BROWSER}   Chrome

*** Test Cases ***
Verify Element Presence
    Open Browser    ${URL}    ${BROWSER}
    Maximize Browser Window
    Sleep    2s

    Page Should Contain Element    id=user-name
    Sleep    3s

    Close Browser
"""

(module4 / "element_verification.robot").write_text(
    element_test,
    encoding="utf-8"
)

result = subprocess.run(
    ["robot", "element_verification.robot"],
    cwd=module4,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print(result.stderr)
```

# Assignment 2: Variables and Data-Driven Automation

Robot Framework supports variables that allow test data to be stored and reused.

Variables are useful for storing:

- URLs
- Browser names
- Usernames
- Passwords
- Other test data

Robot Framework can also perform data-driven testing where the same test is executed using multiple sets of test data.

## Question 4: Using Variables

### Explanation

Robot Framework variables are defined inside the `*** Variables ***` section.

In this example, the username and password are stored as variables.

The variables are then passed to the `Input Text` keyword during the login process.

This makes the test easier to maintain because the test data can be changed in one place.

```python
variables_test = """*** Settings ***
Library    SeleniumLibrary

*** Variables ***
${URL}       https://www.saucedemo.com/
${BROWSER}   Chrome
${USERNAME}  standard_user
${PASSWORD}  secret_sauce

*** Test Cases ***
Login Using Variables
    Open Browser    ${URL}    ${BROWSER}
    Maximize Browser Window
    Sleep    2s

    Input Text    id=user-name    ${USERNAME}
    Sleep    2s

    Input Text    id=password    ${PASSWORD}
    Sleep    2s

    Click Button    id=login-button
    Sleep    4s

    Page Should Contain    Products
    Sleep    3s

    Close Browser
"""

(module4 / "variables_test.robot").write_text(
    variables_test,
    encoding="utf-8"
)

result = subprocess.run(
    ["robot", "variables_test.robot"],
    cwd=module4,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print(result.stderr)
```

## Question 5: Data-Driven Automation

### Explanation

Data-driven automation allows the same test case to execute with different sets of input data.

Instead of writing separate test cases for different users, the test data is stored in an external CSV file.

The DataDriver library reads the CSV file and provides the data to the Robot Framework test.

In this example, multiple valid SauceDemo users are tested using the same login test.

```python
csv_data = """${username},${password}
standard_user,secret_sauce
problem_user,secret_sauce
performance_glitch_user,secret_sauce
"""

(module4 / "test_data.csv").write_text(
    csv_data,
    encoding="utf-8"
)

print("test_data.csv created successfully.")
print((module4 / "test_data.csv").read_text())
```

```python
data_driven_test = """*** Settings ***
Library    SeleniumLibrary
Library    DataDriver    file=test_data.csv
Test Template    Login Test

*** Variables ***
${URL}       https://www.saucedemo.com/
${BROWSER}   Chrome

*** Test Cases ***
Login Test
    ${username}    ${password}

*** Keywords ***
Login Test
    [Arguments]    ${username}    ${password}

    Open Browser    ${URL}    ${BROWSER}
    Maximize Browser Window
    Sleep    2s

    Input Text    id=user-name    ${username}
    Sleep    1s

    Input Text    id=password    ${password}
    Sleep    1s

    Click Button    id=login-button
    Sleep    4s

    Page Should Contain    Products
    Sleep    3s

    Close Browser
"""

(module4 / "data_driven.robot").write_text(
    data_driven_test,
    encoding="utf-8"
)

result = subprocess.run(
    ["robot", "data_driven.robot"],
    cwd=module4,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print(result.stderr)
```

# Assignment 3: Custom Keywords and Libraries

Robot Framework allows users to create their own keywords.

Custom keywords can be created using Python.

A Python class can be imported into Robot Framework as a library.

In this example, a custom keyword is created to calculate the sum of two numbers.

## Question 6: Create a Custom Robot Framework Keyword Using Python

### Explanation

A Python class can be used as a Robot Framework library.

Methods inside the Python class can be called as Robot Framework keywords.

The following Python method calculates the sum of two numbers.

The keyword will be named `Calculate Sum` when used in Robot Framework.

```python
custom_python = """class CustomKeywords:

    def calculate_sum(self, a, b):
        return int(a) + int(b)
"""

(module4 / "custom_keywords.py").write_text(
    custom_python,
    encoding="utf-8"
)

print("custom_keywords.py created successfully.")
```

### Using the Custom Python Library

The Python file is imported using the `Library` setting.

The `Calculate Sum` keyword can then be used directly inside the Robot Framework test case.

The result is verified using the `Should Be Equal As Integers` keyword.

```python
custom_robot_test = """*** Settings ***
Library    custom_keywords.py

*** Test Cases ***
Calculate Sum Using Custom Keyword
    ${result}=    Calculate Sum    10    20
    Log    The calculated sum is ${result}
    Should Be Equal As Integers    ${result}    30
"""

(module4 / "custom_keyword_test.robot").write_text(
    custom_robot_test,
    encoding="utf-8"
)

result = subprocess.run(
    ["robot", "custom_keyword_test.robot"],
    cwd=module4,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print(result.stderr)
```

## Additional Custom Keyword Demonstration

The custom keyword can be reused with different numbers.

This demonstrates that a single custom keyword can perform the same operation for multiple inputs.

```python
custom_robot_multiple = """*** Settings ***
Library    custom_keywords.py

*** Test Cases ***
Calculate Multiple Sums
    ${result1}=    Calculate Sum    10    20
    Should Be Equal As Integers    ${result1}    30

    ${result2}=    Calculate Sum    25    15
    Should Be Equal As Integers    ${result2}    40

    ${result3}=    Calculate Sum    100    50
    Should Be Equal As Integers    ${result3}    150

    Log    All calculations completed successfully
"""

(module4 / "custom_keyword_multiple.robot").write_text(
    custom_robot_multiple,
    encoding="utf-8"
)

result = subprocess.run(
    ["robot", "custom_keyword_multiple.robot"],
    cwd=module4,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print(result.stderr)
```

# Test Reports

Robot Framework automatically generates test execution reports.

The generated files include:

- `output.xml`
- `log.html`
- `report.html`

`report.html` provides a summary of passed and failed test cases.

`log.html` provides detailed execution information for individual keywords.

```python
result = subprocess.run(
    [
        "robot",
        "--output", "output.xml",
        "--log", "log.html",
        "--report", "report.html",
        "."
    ],
    cwd=module4,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print(result.stderr)
```
